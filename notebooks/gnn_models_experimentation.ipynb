### Graph Convolutional Neural Network (GCN)

```python
import torch

from interface.models.gnn import GCNClassifier

# 1. Data Configuration (Simulating 10 nodes)
# These values ​​are set according to the shape of the dataset
# If it is (116, 200), then: Nodes = 116, Features = 200
num_nodes = 360
num_features = 360 # Imagine that each node has 360 initial metrics

# Features (X): 360 nodes, each with 360 values
x = torch.randn(num_nodes, num_features)

# Connections (Edge Index): Defining who connects with whom
# Ex: Node 0-1, 1-2, 2-3...
edge_index = torch.tensor([
    [0, 1, 1, 2, 2, 3, 3, 4, 4, 5], # Origin
    [1, 0, 2, 1, 3, 2, 4, 3, 5, 4]  # Destination
], dtype=torch.long)

# 2. Instantiating the Model
model = GCNClassifier(
    in_channels=num_features, # Input has size 3
    hidden_channels=16,       # Inner layer has size 16
    out_channels=1            # Output has size 1 (probability)
)

print("--- Model Architecture ---")
print(model)

# 3. Performing the Prediction (Inference)
model.eval() # Sets the model to evaluation mode (turns off dropout)
with torch.no_grad():
    output = model(x, edge_index)

print("\n--- GCN Output (Priors for solver) ---")
# .squeeze() removes the extra dimension (transforms [[0.1], [0.5]] into [0.1, 0.5])
priors = output.squeeze().tolist() 

for i, p in enumerate(priors):
    print(f"Node {i}: {p:.4f} (Chance of being 1)")
```

### Graph Attention Network (GAT)

```python
import torch
from interface.models.gnn import GATClassifier


# 1. Configuration (Connectome Example)
NUM_NODES = 10
NUM_FEATURES = 5  # Ex: 5 metrics per region
HEADS = 4         # The network will "look" at the graph in 4 different ways

# Creating simulated data
x = torch.randn(NUM_NODES, NUM_FEATURES)

# Creating random connections (Sparse graph)
edge_index = torch.randint(0, NUM_NODES, (2, 20)) # 20 random edges

# 2. Instantiating the GAT
print(f"Initializing GAT with {HEADS} attention heads...")
model = GATClassifier(
    in_channels=NUM_FEATURES,
    hidden_channels=8, # Each header will have size 8
    out_channels=1,
    heads=HEADS
)

# Architecture Debugging
print(model)

# 3. Running the Prediction
model.eval()
with torch.no_grad():
    output = model(x, edge_index)
    
print("\n--- Result ---")
print(f"Input Shape: {x.shape}")
print(f"Output Shape:   {output.shape} (Expected: [{NUM_NODES}, 1])")

# Extracting to a list (solver format)
priors = output.squeeze().tolist()
print(f"Example of Priors: {priors[:3]}")
```
